```python
from datasets import load_from_disk, Dataset
import pyarrow as pa
import ast
# Option 1: load the whole dataset split folder
ds = load_from_disk('/Users/ghulammustafa/Desktop/Projects/Foundation/MLFoundations/RAG/rag-instruct/train')
df = ds.to_pandas()
```

```python
all_docs = []
for idx, row in df.iterrows():
    docs = row['documents']
    if isinstance(docs, str):
        docs = ast.literal_eval(docs)  # parse if stored as string
    for doc in docs:
        all_docs.append({
            "text": doc,
            "question": row['question'],   # optional metadata
            "answer": row['answer']        # optional metadata
        })

print(f"Total raw docs: {len(all_docs)}")
print(all_docs[:1] )
 
```
